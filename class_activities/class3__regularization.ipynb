```python
import pandas as pd
import os
ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(os.path.join(ROOT, "Datasets", "Melbourne.csv"))
df_copy = df.copy()
pd.set_option("display.max_columns", None)
df.head()
```

## Data Science

```python
df.shape
```

```python
cols=['Suburb', 'Rooms','Type','Method','SellerG','Regionname','Propertycount', 'Distance', 'CouncilArea','Bedroom2', 'Bathroom', 'Car', 'Landsize', 'Landsize','BuildingArea', 'Price']
df = df[cols]
df.shape
```

```python
df.nunique()
```

```python
df.isna().sum()
```

```python
df["Distance"] = df["Distance"].fillna(df["Distance"].mean())
df["Propertycount"] = df["Propertycount"].fillna(df["Propertycount"].mean())
df["Landsize"] = df["Landsize"].fillna(df["Landsize"].mean())
df["BuildingArea"] = df["BuildingArea"].fillna(df["BuildingArea"].mean())
df["Car"] = df["Car"].fillna(0)
df["Bathroom"] = df["Bathroom"].fillna(0)
df["Bedroom2"] = df["Bedroom2"].fillna(0)

df.isna().sum()
```

```python
df.dropna(axis=0,inplace=True) # Axis de lo que quiero eliminar y Inplace=True => Actualizar
df.isna().sum()
```

## One hot enconding:

```python
df.shape
```

```python
df = pd.get_dummies(df, drop_first = True)
```

```python
df.shape
```

## Machine Learning

```python
# Dividir muestras a features y labe
x = df.drop("Price", axis = 1)
y = df["Price"]

y.head()
```

```python
# Dividir muestras a entrenamiento y prueba
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

### Modelo de Regresión Lineal

```python
# Entrenamiento
LR_model = LinearRegression()
LR_model.fit(x_train, y_train)
```

```python
# Validación de entrenamiento
y_pred_train = LR_model.predict(x_train)
print(f"R2 Score: {r2_score(y_train, y_pred_train)}")
```

```python
# Testeo
y_pred_test = LR_model.predict(x_test)
print(f"R2 Score: {r2_score(y_test, y_pred_test)}")
```

## Rgularización

### Regularización con Lasso

```python
L1 = Lasso(alpha = 50, max_iter = 100, tol = 0.1)
L1.fit(x_train, y_train)
```

```python
y_pred_L1 = L1.predict(x_test)
print(f"R2 Score: {r2_score(y_test, y_pred_L1)}")
```

### Regularización con Ridge

```python
L2 = Ridge(alpha = 50, max_iter = 100, tol = 0.1)
L2.fit(x_train, y_train)
```

```python
y_pred_L2 = L2.predict(x_test)
print(f"R2 Score: {r2_score(y_test, y_pred_L2)}")
```

### Regularización con ElasticNet

```python
L12 = ElasticNet(alpha = 0.5, max_iter = 100, l1_ratio = 0.5, tol = 0.1)
L12.fit(x_train, y_train)
```

```python
y_pred_L12 = L12.predict(x_test)
print(f"R2 Score: {r2_score(y_test, y_pred_L12)}")
```
