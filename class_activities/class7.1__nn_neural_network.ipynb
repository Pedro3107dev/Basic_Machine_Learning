```python
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
import tensorflow as tf
```

### Data science

```python
digits = load_digits()
digits.data.shape
```

```python
print(digits.data)
```

```python
plt.figure(figsize = (10, 8))

for i, (img, label) in enumerate(zip(digits.images[0:6], digits.target[0:6])):
    plt.subplot(1, 6, i + 1)
    plt.imshow(img, cmap = 'gray')
    plt.title(f'Label: {label}')
    plt.axis('off')
```

### Create datasets

```python
x_train, x_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size = 0.2, random_state = 42)
```

### Red Neuronal

```python
num_features = len(digits.data[0])
num_classes = len(digits.target_names)

NN_model = tf.keras.Sequential([
    tf.keras.layers.Dense(256, activation = "relu", input_shape = (num_features,)),
    tf.keras.layers.Dense(128, activation = "relu"),
    tf.keras.layers.Dense(64, activation = "relu"),
    tf.keras.layers.Dense(32, activation = "relu"),
    tf.keras.layers.Dense(num_classes, activation = "softmax")
])
NN_model.compile(optimizer = "adam", loss = "sparse_categorical_crossentropy", metrics = ["accuracy"])
```

### Training

```python
# Train the model
NN_model.fit(x_train, y_train, epochs = 10)
```

### Testing

```python
# Test the model
loss, accuracy = NN_model.evaluate(x_test, y_test)
print(f"Test Accuracy: {accuracy}")
```
